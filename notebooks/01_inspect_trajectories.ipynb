Inspect the raw `points.csv`, cleaned trajectory file, yaw/speed derivation, and basic trajectory statistics.

```python
from pathlib import Path
import pandas as pd
import numpy as np

ROOT = Path('..') if Path.cwd().name == 'notebooks' else Path('.')
points_path = ROOT / 'data/raw/trajectories/points.csv'
clean_path = ROOT / 'data/interim/trajectory_clean.parquet'

points_path, clean_path
```

```python
points = pd.read_csv(points_path)
print(points.shape)
points.head()
```

```python
print(points.dtypes)
print(points['id'].nunique())
print(points.groupby('id').size().describe())
```

```python
if clean_path.exists():
    traj = pd.read_parquet(clean_path)
else:
    traj = None
    print('Run scripts/01_prepare_trajectories.py first.')
traj.head() if traj is not None else None
```

```python
if traj is not None:
    display(traj[['trajectory_id','t','timestamp','lon','lat','x','y','yaw_deg','speed_mps']].head())
    print(traj[['speed_mps','yaw_deg','dt_next_s','step_next_m']].describe())
```

```python
if traj is not None:
    import matplotlib.pyplot as plt
    counts = traj.groupby('trajectory_id').size()
    counts.hist(bins=30)
    plt.title('Points per trajectory')
    plt.xlabel('points')
    plt.ylabel('trajectory count')
    plt.show()
```
