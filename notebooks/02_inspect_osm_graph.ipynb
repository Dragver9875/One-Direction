Inspect the processed OSM road graph, edge table, node table, and road-class distribution.

```python
from pathlib import Path
import pandas as pd

ROOT = Path('..') if Path.cwd().name == 'notebooks' else Path('.')
edge_path = ROOT / 'data/processed/road_graph/edge_table.parquet'
node_path = ROOT / 'data/processed/road_graph/node_table.parquet'
report_path = ROOT / 'data/processed/road_graph/osm_graph_report.json'
edge_path, node_path
```

```python
if edge_path.exists():
    edges = pd.read_parquet(edge_path)
    print(edges.shape)
    display(edges.head())
else:
    edges = None
    print('Run scripts/03_build_osm_graph.py first.')
```

```python
if edges is not None:
    display(edges['road_class'].value_counts())
    display(edges[['length_m','bearing_rad']].describe())
```

```python
if node_path.exists():
    nodes = pd.read_parquet(node_path)
    print(nodes.shape)
    display(nodes.head())
else:
    nodes = None
```

```python
if edges is not None:
    import matplotlib.pyplot as plt
    edges['length_m'].clip(upper=500).hist(bins=50)
    plt.title('Road-segment length distribution, clipped at 500 m')
    plt.xlabel('length_m')
    plt.ylabel('count')
    plt.show()
```
