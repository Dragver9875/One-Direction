```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path('..') if Path.cwd().name == 'notebooks' else Path('.')
metrics_path = ROOT / 'outputs/metrics/gnn_hmm_metrics.json'
error_path = ROOT / 'outputs/metrics/error_cases.csv'
match_path = ROOT / 'outputs/matches/gnn_hmm_matches.parquet'
```

```python
if metrics_path.exists():
    print(json.dumps(json.loads(metrics_path.read_text()), indent=2))
else:
    print('Run scripts/09_evaluate.py first.')
```

```python
if match_path.exists():
    matches = pd.read_parquet(match_path)
    print(matches.shape)
    display(matches.head())
else:
    matches = None
    print('No match file found yet.')
```

```python
if error_path.exists():
    errors = pd.read_csv(error_path)
    print(errors.shape)
    display(errors.head())
else:
    errors = None
    print('No error case file found yet.')
```

```python
if errors is not None and len(errors) > 0:
    display(errors.groupby('trajectory_id').size().sort_values(ascending=False).head(20))
```

```python
if matches is not None and 'confidence' in matches.columns:
    import matplotlib.pyplot as plt
    matches['confidence'].dropna().hist(bins=50)
    plt.title('Prediction confidence distribution')
    plt.xlabel('confidence')
    plt.ylabel('count')
    plt.show()
```
